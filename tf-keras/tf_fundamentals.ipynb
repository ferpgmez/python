```python
rank_4_tensor = tf.zeros(shape=[2, 3, 4, 5])
rank_4_tensor
```

```python
print("Tipo de dato de cada elemento:", rank_4_tensor.dtype)
print("Numero de dimensiones(rango)", rank_4_tensor.ndim)
print("Forma del tensor:", rank_4_tensor.shape)
print("Numero de elementos:", tf.size(rank_4_tensor).numpy())
print("Elementos en el primer eje del tensor", rank_4_tensor.shape[0])
print("Elememtos en el ultimo eje del tensor", rank_4_tensor.shape[-1])
```

```python
# obtenemos los dos primeros items de cada dimension
rank_4_tensor[:2, :2, :2, :2]
```

```python
# obtenemos los dos primeros items de cada dimension excepto el ultimo
rank_4_tensor[:1, :1, :1, :]
```

```python
#creamos un tensor de rango2
rank_2_tensor = tf.constant([[10, 7], [3, 4]])
#obtenemos el ultimo item de cada fila
rank_2_tensor[:, -1]
```

```python
# anhadimos una dimension extra (al final)
rank_3_tensor = rank_2_tensor[..., tf.newaxis]
rank_2_tensor, rank_3_tensor
```

```python
#podemos hacer lo mismo usando tf.expand_dims
rank_3_tensor = tf.expand_dims(rank_2_tensor, axis=-1)
rank_2_tensor, rank_3_tensor
```

```python
X = tf.constant([[1, 2],
                 [3, 4],
                 [5, 6]])
Y = tf.constant([[7, 8],
                 [9, 10],
                 [11, 12]])
X_tr = tf.transpose(X)
X_tr @ Y
```

```python
Y_tr = tf.transpose(Y)
Y_tr
```

```python
X @ Y_tr
```
